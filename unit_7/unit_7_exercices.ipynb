### Quick check 31.4
Suppose you create a Door object type in the following way:

```python
class Door(object):
    def __init__(self):
        self.width=1
        self.height=1
        self.open=False
    def change_state(self):
        self.open=not self.open
    def scale(self, factor):
        self.height *= factor
        self.width *=factor
```

```python
square_door=Door()
```

```python
square_door.change_state()
```
